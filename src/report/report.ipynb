```python
import os
import json
import sys

import time
def read_json(path):
    with open(path, 'r', encoding="utf-8") as f:
        data = json.load(f)
    return data

def write_json(data, path):
    
    if not os.path.exists(os.path.dirname(path)):
        os.makedirs(os.path.dirname(path))
    with open(path, 'w', encoding="utf-8") as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
fewrel_report = "../converse_relations/results/fewrel/"
report = []
for exp in ['MT', 'synthetic','original']:

    for model in os.listdir(os.path.join(fewrel_report, exp)):
        if model != "reports" and model !=".DS_Store":
            for input_file in os.listdir(os.path.join(fewrel_report, exp, 'reports', model)):
                if input_file.endswith(".json"):
                    print(f"Model: {model}, Experiment: {exp}, Input File: {input_file}")
                    file_path = os.path.join(fewrel_report, exp, 'reports', model, input_file)
                    print(f"Reading file: {file_path}")
                    data = read_json(file_path)
                    print(data)
                    result = {
                        "model": model,
                        "experiment": exp,
                        "input_file": '_'.join(input_file.split('_')[:-2])+".json",
                        "accuracy": data.get("accuracy"),
                        "micro_precision": data.get("micro_precision"),
                        "micro_recall": data.get("micro_recall"),
                        "micro_f1": data.get("micro_f1")
                    }
                    report.append(result)
```

```python
import pandas as pd


df = pd.DataFrame(report)

cols_to_agg = ["accuracy", "micro_precision", "micro_recall", "micro_f1"]
df[cols_to_agg] = df[cols_to_agg].apply(pd.to_numeric)

summary_df = df.groupby(['model', 'input_file', 'experiment'])[cols_to_agg].mean()

summary_df.to_csv("./converse_relations/results/fewrel/summary_report.csv")
```
